```python
import requests
import base64
import json

def encode_image_to_base64(image_path):
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def ask_llava(image_path, prompt):
    url = "http://localhost:11434/api/generate"
    encoded_image = encode_image_to_base64(image_path)

    payload = {
        "model": "llava",
        "prompt": prompt,
        "images": [encoded_image],
        "stream": True
    }

    response = requests.post(url, json=payload, stream=True)
    response.raise_for_status()

    output = ""
    for line in response.iter_lines():
        if line:
            try:
                data = json.loads(line.decode("utf-8"))
                output += data.get("response", "")
            except json.JSONDecodeError:
                continue

    return output.strip()

prompt = """
Summarise the image in 3–5 clear sentences as if explaining it to someone who cannot see it. Focus on:

- What the image shows or represents
- Key elements and how they relate
- Any visible text, labels, or symbols
- Its likely purpose or use
- A few relevant keywords for natural language search

Keep it concise, informative, and suitable for semantic vector search.
"""

result = ask_llava("c.jpg", prompt)
print("\nLLaVA Output:\n", result)
```
